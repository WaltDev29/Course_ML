# 주성분 분석

## 1. 패키지 설정

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. 데이터 작성

```python
# 원데이터
X = np.array([[50,73], [65,75], [75,80], [80,82], [95,85]])
print(X)
```

## 3. 탐색적 데이터 분석

```python
# 원데이터의 1열(X1)과 2열(X2)의 산포도
plt.scatter(X[:,0], X[:,1])
plt.xlabel('X1')
plt.ylabel('X2')
plt.xlim(0, 100)
plt.ylim(0, 100)

plt.show()
```

```python
#열별 평균과 분산
print(np.mean(X, axis=0))
print(np.var(X, axis=0))
```

## 4. 데이터 스케일링

```python
# 데이터 표준화
scalerX = StandardScaler()
scalerX.fit(X)
X_std = scalerX.transform(X)
print(X_std)
```

```python
# 표준화된 데이터의 1열(X1)과 2열(X2)의 산포도
plt.scatter(X_std[:,0], X_std[:,1])
plt.xlabel('X1')
plt.ylabel('X2')
plt.xlim(-3, 3)
plt.ylim(-3, 3)
```

```python
# 열별 평균 및 모공분산행렬
print(np.mean(X_std, axis=0))
print(np.cov(X_std[:,0], X_std[:,1], ddof=0))
```

## 5. 주성분 분석

```python
# 주성분의 수 설정 및 주성분분석
pca = PCA(n_components=2)
pca.fit(X_std)
```

## 6. 주성분 추출

```python
# 주성분 분산 및 분산비율
print(pca.explained_variance_)
print(pca.explained_variance_ratio_)
```

## 7. 주성분으로의 데이터 변환

```python
# 주성분으로 데이터 변환
Z = pca.fit_transform(X_std)
print(Z)
```

```python
# 주성분 좌표계의 분포
plt.scatter(Z[:,0], Z[:,1])
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.xlim(-2.5, 2.5)
plt.ylim(-2.5, 2.5)

plt.show()
```

## 8. 시각화 해석

```python
# 주성분 로딩(고유벡터) 확인
loadings = pca.components_
print(loadings)
```

```python
# 주성분 로딩 분포 시각화
plt.scatter(loadings[:,0], loadings[:,1], color='w')
plt.xlabel('PC1')
plt.ylabel('PC2')

plt.scatter(Z[:,0], Z[:,1])
plt.scatter(X_std[:,0], X_std[:,1])
plt.xlim(-3, 3)
plt.ylim(-3, 3)

# 행과 열의 수 및 이름 설정
rows, columns = loadings.shape
rows_names = ['x1', 'x2']
for i in range(rows):
    # 선 긋기
    plt.arrow(0, 0, loadings[i,0], loadings[i,1], color='r', alpha=0.5)
    # 변수명 출력
    plt.text(loadings[i,0]*1.2, loadings[i,1]*1.2, rows_names[i], color='g', ha='center', va='center')
```
